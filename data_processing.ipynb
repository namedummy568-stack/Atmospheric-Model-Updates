```python
import pandas as pd

# Sample data processing code
data = {'Date': pd.to_datetime(['2023-01-01', '2023-01-02', '2023-01-03']),
        'Temperature': [273.15, 275.0, 272.5],
        'Pressure': [1000, 998, 1001]}
df = pd.DataFrame(data)

print("Initial Data:")
print(df.head())

# Example: Convert Temperature from Kelvin to Celsius
df['Temperature_Celsius'] = df['Temperature'] - 273.15

print("
Data with Temperature in Celsius:")
print(df.head())
```
